```python
import pandas as pd
from sqlalchemy import create_engine, text
data = pd.read_csv(r"Project_files\G21_Logistics – Shipments  Carriers with Warehouses_merged.csv")
```

```python
columns_to_check = ['shipment_id', 'carrier_id', 'warehouse_id']
data = data.drop_duplicates(subset=columns_to_check, keep='first')
data['id'] = range(1, len(data) + 1)
data
```

```python
# Define as colunas a importar
columns_to_import = ['warehouse_id', 'location', 'capacity', 'latitude', 'longitude']

# Remove duplicados apenas nessas colunas
df_selected = data[columns_to_import].drop_duplicates()
df = df_selected.drop_duplicates()

# Ordena por warehouse_id
df = df.sort_values(by='warehouse_id')

# Caminho e engine
database_path = r'data\App.db'
engine = create_engine(f'sqlite:///{database_path}')
table_name = 'Warehouses'

# Cria a tabela (corrige a vírgula em capacity INTEGER!)
with engine.connect() as conn:
    conn.execute(text("DROP TABLE IF EXISTS Warehouses;"))
    conn.execute(text("""
        CREATE TABLE Warehouses (
            warehouse_id INTEGER PRIMARY KEY,
            location TEXT,
            capacity INTEGER,
            latitude REAL,
            longitude REAL
        );
    """))
    conn.commit()

# Insere os dados
df.to_sql(table_name, con=engine, if_exists='replace', index=False)
```

```python
columns_to_import = ['shipment_id', 'status', 'origin', 'destination', 'shipment_date', 'tracking_number']
df_selected = data[columns_to_import]
database_path = r'data\App.db'
engine = create_engine(f'sqlite:///{database_path}')
table_name = 'Shipments'

# 1. Cria a tabela manualmente com a primary key
with engine.connect() as conn:
    conn.execute(text("DROP TABLE IF EXISTS Shipments;"))
    conn.execute(text("""
        CREATE TABLE Shipments (
            shipment_id INTEGER PRIMARY KEY,
            status TEXT,
            origin TEXT,
            destination TEXT,
            shipment_date TEXT,
            tracking_number TEXT
        );
    """))
    conn.commit()

# 2. Insere os dados
df_selected.to_sql(table_name, con=engine, if_exists='replace', index=False)
```
