```python
import pandas as pd
from sqlalchemy import create_engine, text
data = pd.read_csv(r"Project_files\G21_Logistics – Shipments  Carriers with Warehouses_merged.csv")
```

```python
columns_to_check = ['shipment_id', 'carrier_id', 'warehouse_id']
data = data.drop_duplicates(subset=columns_to_check, keep='first')
data['id'] = range(1, len(data) + 1)
data
```

```python
# Define as colunas a importar
columns_to_import = ['warehouse_id', 'location', 'capacity', 'latitude', 'longitude']

# Remove duplicados apenas nessas colunas
df_selected = data[columns_to_import].drop_duplicates()
df = df_selected.drop_duplicates(subset='warehouse_id')

# Ordena por warehouse_id
df = df.sort_values(by='warehouse_id')

# Caminho e engine
database_path = r'data\App.db'
engine = create_engine(f'sqlite:///{database_path}')
table_name = 'Warehouses'

# Cria a tabela (corrige a vírgula em capacity INTEGER!)
with engine.connect() as conn:
    conn.execute(text("DROP TABLE IF EXISTS Warehouses;"))
    conn.execute(text("""
        CREATE TABLE Warehouses (
            warehouse_id INTEGER PRIMARY KEY,
            location TEXT,
            capacity INTEGER,
            latitude REAL,
            longitude REAL
        );
    """))
    conn.commit()

# Insere os dados
df.to_sql(table_name, con=engine, if_exists='replace', index=False)
```

```python
# Define as colunas a importar
columns_to_import = ['shipment_id', 'status', 'origin', 'destination', 'shipment_date', 'tracking_number']

# Remove duplicados apenas nessas colunas
df_selected = data[columns_to_import].drop_duplicates()
df_shipments = df_selected.drop_duplicates(subset='shipment_id')

# Ordena por shipment_id
df_shipments = df_shipments.sort_values(by='shipment_id')

# Caminho e engine
database_path = r'data\App.db'
engine = create_engine(f'sqlite:///{database_path}')
table_name = 'Shipments'

# Cria a tabela manualmente com a primary key
with engine.connect() as conn:
    conn.execute(text("DROP TABLE IF EXISTS Shipments;"))
    conn.execute(text("""
        CREATE TABLE Shipments (
            shipment_id INTEGER PRIMARY KEY,
            status TEXT,
            origin TEXT,
            destination TEXT,
            shipment_date TEXT,
            tracking_number TEXT
        );
    """))
    conn.commit()

# Insere os dados
df_shipments.to_sql(table_name, con=engine, if_exists='replace', index=False)
```

```python
columns_to_import = ['carrier_id', 'vehicle_type', 'capacity', 'name']

# Remove duplicados apenas nessas colunas
df_selected = data[columns_to_import].drop_duplicates()
df_carriers = df_selected.drop_duplicates(subset='carrier_id')

# Ordena por carrier_id
df_carriers = df_carriers.sort_values(by='carrier_id')

# Caminho e engine
database_path = r'data\App.db'
engine = create_engine(f'sqlite:///{database_path}')
table_name = 'Carriers'

# Cria a tabela manualmente com a primary key
with engine.connect() as conn:
    conn.execute(text("DROP TABLE IF EXISTS Carriers;"))
    conn.execute(text("""
        CREATE TABLE Carriers (
            carrier_id INTEGER PRIMARY KEY,
            vehicle_type TEXT,
            capacity INTEGER,
            name TEXT
        );
    """))
    conn.commit()

# Insere os dados
df_carriers.to_sql(table_name, con=engine, if_exists='replace', index=False)
```

```python
columns_to_import = ['id', 'shipment_id', 'carrier_id']

# Remove duplicados apenas nessas colunas
df_selected = data[columns_to_import].drop_duplicates()
df_shipment_details = df_selected.drop_duplicates(subset='id')

# Ordena por id
df_shipment_details = df_shipment_details.sort_values(by='id')

# Caminho e engine
database_path = r'data\App.db'
engine = create_engine(f'sqlite:///{database_path}')
table_name = 'Shipment_details'

# Insere os dados
df_shipment_details.to_sql(table_name, con=engine, if_exists='replace', index=False)
```
